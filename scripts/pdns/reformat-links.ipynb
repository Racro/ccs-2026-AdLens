```python

import json

def read_cat_links(cat_links_path):
    # read from JSON file
    distinct_domains = set()
    cat_domains = json.loads(open(cat_links_path).read())
    for ad_category, domains in cat_domains.items():
        distinct_domains.update(domains)
        print(f"Category: {ad_category}, Number of links: {len(domains)}")
    return distinct_domains

def main():
    distinct_domains = read_cat_links('unique_domains_all.json')
    with open('all_ad_link_domains.txt', 'w') as f:
        for domain in distinct_domains:
            f.write(f"{domain}\n")

    print(f"Total distinct URLs: {len(distinct_domains)}")

main()
```
